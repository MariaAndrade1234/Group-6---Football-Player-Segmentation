```python
# %% [markdown]
# # Validação do Dataset
# 
# Este notebook valida a integridade e qualidade do dataset convertido

# %%
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

# %%
# Configurações
dataset_dir = "../data/yolo_dataset"
images_dir = os.path.join(dataset_dir, "images", "train")
labels_dir = os.path.join(dataset_dir, "labels", "train")

# %%
# Verificar correspondência entre imagens e labels
image_files = set([f.split('.')[0] for f in os.listdir(images_dir) if f.endswith('.jpg')])
label_files = set([f.split('.')[0] for f in os.listdir(labels_dir) if f.endswith('.txt')])

missing_labels = image_files - label_files
extra_labels = label_files - image_files

print(f"Total images: {len(image_files)}")
print(f"Total labels: {len(label_files)}")
print(f"Images missing labels: {len(missing_labels)}")
print(f"Extra labels without images: {len(extra_labels)}")

# %%
# Visualizar algumas anotações
def plot_annotations(img_path, label_path):
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img.shape[:2]
    
    with open(label_path, 'r') as f:
        lines = f.readlines()
    
    for line in lines:
        cat_id, xc, yc, bw, bh = map(float, line.strip().split())
        x1 = int((xc - bw/2) * w)
        y1 = int((yc - bh/2) * h)
        x2 = int((xc + bw/2) * w)
        y2 = int((yc + bh/2) * h)
        
        cv2.rectangle(img, (x1, y1), (x2, y2), (255, 0, 0), 2)
        cv2.putText(img, str(int(cat_id)), (x1, y1-5), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 1)
    
    plt.figure(figsize=(10, 10))
    plt.imshow(img)
    plt.axis('off')
    plt.show()

# %%
# Visualizar 5 amostras aleatórias
sample_files = np.random.choice(list(image_files), 5, replace=False)
for file in sample_files:
    img_path = os.path.join(images_dir, file + '.jpg')
    label_path = os.path.join(labels_dir, file + '.txt')
    plot_annotations(img_path, label_path)
```
