```python
# %% [markdown]
# # Avaliação do Modelo
# 
# Este notebook avalia o modelo YOLOv8 treinado

# %%
from scripts.evaluate_model import evaluate_model

# %%
# Configurações
model_path = "runs/detect/football_yolov8m/weights/best.pt"
data_yaml = "../data/yolo_dataset/data.yaml"

# %%
# Avaliar modelo
metrics = evaluate_model(model_path, data_yaml, split='val', conf=0.5)

# %%
# Visualizar predições
from ultralytics import YOLO
import glob
import matplotlib.pyplot as plt

model = YOLO(model_path)
sample_images = glob.glob("../data/yolo_dataset/images/val/*.jpg")[:5]

for img_path in sample_images:
    results = model.predict(img_path, conf=0.5)
    for r in results:
        im_array = r.plot()  # plot a BGR numpy array of predictions
        plt.figure(figsize=(10, 10))
        plt.imshow(im_array[..., ::-1])  # RGB PIL image
        plt.axis('off')
        plt.show()
```
