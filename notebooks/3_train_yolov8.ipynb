```python
# %% [markdown]
# # Treinamento do YOLOv8
# 
# Este notebook treina um modelo YOLOv8 no dataset de futebol

# %%
from scripts.train_model import train_yolov8
import yaml

# %%
# Criar data.yaml
dataset_dir = "../data/yolo_dataset"
data_yaml = os.path.join(dataset_dir, "data.yaml")

with open(data_yaml, 'w') as f:
    yaml.dump({
        'path': dataset_dir,
        'train': 'images/train',
        'val': 'images/val',
        'names': {i: name for i, name in enumerate(open(os.path.join(dataset_dir, 'classes.txt')).read().splitlines())}
    }, f)

# %%
# Treinar modelo
results = train_yolov8(data_yaml, model_size='m', epochs=100, imgsz=640, batch=16)

# %%
# Visualizar métricas de treino
import pandas as pd
import matplotlib.pyplot as plt

metrics_file = "runs/detect/football_yolov8m/results.csv"
metrics = pd.read_csv(metrics_file)

plt.figure(figsize=(12, 8))
plt.plot(metrics['epoch'], metrics['metrics/precision(B)'], label='Precision')
plt.plot(metrics['epoch'], metrics['metrics/recall(B)'], label='Recall')
plt.plot(metrics['epoch'], metrics['metrics/mAP50(B)'], label='mAP50')
plt.plot(metrics['epoch'], metrics['metrics/mAP50-95(B)'], label='mAP50-95')
plt.xlabel('Epoch')
plt.ylabel('Metric Value')
plt.title('Training Metrics')
plt.legend()
plt.grid()
plt.show()
```
